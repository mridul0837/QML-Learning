## Qiskit Version

```python
from qiskit import QuantumCircuit

qc = QuantumCircuit(1, 1)

qc.h(0)
qc.measure(0, 0)

print(qc)
```

```python
qc1 = QuantumCircuit(1, 1)
qc1.h(0)
qc1.h(0)
qc1.measure(0,0)

print(qc1)
```

```python
# Circuit with 1 qubit and 1 classical bit
qc2 = QuantumCircuit(1, 1)

# Apply Pauli Gates
qc2.x(0)  # Pauli-X (Bit-flip)
qc2.y(0)  # Pauli-Y (Bit + Phase-flip)
qc2.z(0)  # Pauli-Z (Phase-flip)

qc2.measure(0, 0)

print(qc2)
```

## PennyLane Version

```python
import pennylane as qml

dev = qml.device("default.qubit", wires=1)

@qml.qnode(dev)
def circuit():
    qml.Hadamard(wires=0)
    return qml.probs(wires=0)

print(circuit())
```

```python
# Define the 1-qubit simulator device
dev1 = qml.device("default.qubit", wires=1)

@qml.qnode(dev1)
def circuit1():
    # Apply double Hadamard gate to qubit (wire) 0
    qml.Hadamard(wires=0)
    qml.Hadamard(wires=0)
    
    return qml.probs(wires=0)

print(circuit1())
```

```python
dev2 = qml.device("default.qubit", wires=1)

@qml.qnode(dev2)
def circuit2():
    # Apply Pauli Gates
    qml.PauliX(wires=0)  # Pauli-X
    qml.PauliY(wires=0)  # Pauli-Y
    qml.PauliZ(wires=0)  # Pauli-Z
    
    return qml.state()

print(circuit2())
```
